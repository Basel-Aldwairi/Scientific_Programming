#### Introduction:
Scientists can simulate how atomic particles behave within a confined region for a given quantum number (i.e., 'n') based on the physics dictated by quantum mechanics. The simplest model for this is called the "particle in a one-dimensional box," where we imagine a particle moving in a single dimension between values of zero and a specified length (i.e., 'L') [1]. The quantum number is roughly understood as an allowed state that the particle exists in, with n=1 being the lowest energy state.
The wavefunction (i.e., $\psi (x)$) is a mathematical expression [2] that contains everything known about a particle when it is described completely. The square of the wavefunction (i.e., $\psi^ {2}(x)$) provides the probability of finding the particle at a specified point in space [1]. For a one-dimensional box, a particle (e.g., an electron) is confined to a single axis (e.g., the 'x'-axis). The equation is given by

$$ \psi(x) = \left(\frac{2}{L}\right)^{\frac{1}{2}} \sin\left(\frac{n\pi x}{L}\right) $$

This model can describe electron behavior in linear conjugated molecules like butadiene, which has an approximate length of $559.0$ pm[3,4].

built-in library imports of math functions and constants, as well as the implementation of the Probability Density Function (PDF)

```python
from math import sin, pi

def probability_density(particle_location : int | float, box_length : int | float, quantum_number : int) -> float:
    '''Wave function of a particle in one-dimensional box

        Args:
            particle_location: location of particle in space (unit = m)
            box_length: Length of the one dimensional box (unit = m)
            quantum_number: The allowed energy state of the particle (n)

        Return:
            Probability of finding the particle at a specific location in space (x)
    '''
    
    if not isinstance(particle_location, (int, float)) or particle_location < 0.0:
        raise ValueError('particle_location must be a positive number.')

    if not isinstance(box_length, (int, float)) or box_length <= 0.0:
        raise ValueError('box_length must be a positive number greater than 0.')

    if not isinstance(quantum_number, int) or quantum_number < 1:
        raise ValueError('quantum_number must be a positive integer.')

    
    coefficient = pow(2/box_length,1/2)
    wave_argument = (quantum_number * pi * particle_location / box_length)

    wave = sin(wave_argument)

    phi = coefficient * wave

    probability = pow(phi, 2)
    
    return probability
```

#### Task 1: 
Probability Densities (PDF) Calculate the probability density, which is the square of the wavefunction , $$\psi^ {2}(x)$$ for an electron in a one-dimensional box of length of $559.0$ pm for quantum numbers n = 1 and n = 2.

---

Calculating the step size in order to calculate the PDF
$$size = {L}/{steps}$$

size : step_size

steps : number_of_steps = $10^{5}$

L : box_length = $559.0$ pm


```python
box_length = 559.0e-12 # meter
number_of_steps = 1e5
step_size = box_length/number_of_steps #meter
```

Generating equally spaced out steps ($x$) and initializing density dictionaries for both $n=1$ and $n=2$

```python
x_list = [step * step_size for step in range(int(number_of_steps+1))]

densities = {
    1:[],
    2:[],
}
```

Calculating Probability Densities for $n=1$ and $n=2$

$$\psi^ {2}(x) $$

```python
for quantum_number in range(1,len(densities)+1):
    for x in x_list:
        
        n_probability = probability_density(particle_location=x, box_length=box_length, quantum_number=quantum_number)
        densities[quantum_number].append(n_probability)
```

Function to show results for Probability Densities for $n=1$ and $n=2$ from the start, middle and end

```python
from typing import List

def table_display(x_list : List[float], density : List[float], quantum_number : int) -> None:
    '''Displays a formatted table summary of calculated probability densities.

    Args:
        x_list : A list of x over the box length
        density : A list of collected probability densities
        quantum_number : The energy state (n) associated with these densities

    Returns:
        None
    '''

    if not isinstance(x_list, List):
        raise ValueError('x_list must be a list of floats')
    if not isinstance(density, List):
        raise ValueError('density must be a list of floats')
    if not isinstance(quantum_number, int) or quantum_number < 1:
        raise ValueError('quantum_number must be a positive integer.')
        

    # Heading
    print(f'{f'Quantum Number : {quantum_number}':^40}')
    print(f'_'*40)
    print()

    # Column Titles
    print(f'{'x (pm)':<7}:{'Phi^2(x)':^14}')

    # Calculating Ranges to view results
    density_length = len(density)
    mid_point = density_length // 2 

    selection_step_size = 100
    number_of_points = 5
    offset_size = selection_step_size * number_of_points
    
    selected_points = [(0,offset_size), 
                       (mid_point - offset_size // 2 - selection_step_size // 2, mid_point + offset_size // 2 + 1), 
                       (density_length - offset_size, density_length)]

    # Print 
    for selected_pair in selected_points:
        print(f'{':':^15}')
        for i in range (selected_pair[0],selected_pair[1],selection_step_size):
            x_m = x_list[i] # meter
            x_pm = x_m * 1e12 # picometer
            print(f'{x_pm:<6.1f} : {density[i]:>14.3e}')

        
    print()

    
```

Following are the calculated probability densities

```python
for quantum_number, density in densities.items():
    table_display(x_list=x_list,density=density, quantum_number=quantum_number)
```

#### Task 2: 

Maximum Probability Length Identify the position (i.e., the length) where the electron has the maximum probability of being located for the first quantum number ($n = 1$).

$$ x = \operatorname*{argmax}_x (\psi^2(x)) $$

```python
max_density_n1 = max(densities[1])
max_density_n1_index = densities[1].index(max_density_n1)
max_density_n1_location = x_list[max_density_n1_index] # meter
max_density_n1_location_pm = max_density_n1_location * 1e12 # meter -> picometer
```

Displaying the results

```python
print(f'Maximum Density at n = 1 : {max_density_n1:0.3e}')
print(f'Location of maximum Density at n = 1 : {max_density_n1_location_pm:0.1f} pm')
```

---

#### Results:
- At $n = 1$, the maximum probability density occurs exactly at the midpoint of the box length.
- At $n = 2$, the probability density at the midpoint drops to zero (a node). Instead of being in the center, the particle is most likely to be found in two separate regions on either side.
